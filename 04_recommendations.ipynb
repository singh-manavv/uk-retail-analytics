## Purpose

This notebook pulls together the findings from the EDA, RFM segmentation, and SQL analysis into concrete business recommendations. Each recommendation is grounded in a specific finding, targets a specific problem, and includes a measurable outcome where possible.

The analysis covered four areas:
- **Revenue and category performance** — where the money comes from and where the margin is
- **Customer segmentation** — who the most and least valuable customers are
- **Loyalty programme effectiveness** — whether the programme is identifying the right customers
- **Channel and regional performance** — where the business is strongest and where it is leaving revenue behind

## Recommendation 1 — Fix the Loyalty Programme

**Finding:** Bronze customers outspend Platinum on average (£663 vs £582). The average RFM score follows the same pattern — Bronze scores 2.52, Platinum scores 2.43, the lowest of any tier. Nearly half of all Champions (840 out of 1,783) hold Bronze status. The programme is rewarding the wrong customers.

**What this means:** Loyalty tier is not being assigned based on actual purchase behaviour. The business's most valuable customers are being treated as its least valued, while Platinum status is going to customers with lower engagement and lower spend.

**Recommendation:** Rebuild the loyalty tier assignment logic using RFM scores rather than whatever metric is currently driving it. Platinum should map to Champions and high-scoring Loyals. Conduct an immediate audit of current Platinum holders — identify which ones are Champions and which are not, and create a fast-track upgrade path for Bronze/Silver Champions.

**Target outcome:** Loyalty tier distribution aligns with RFM segment distribution within two programme cycles.

## Recommendation 2 — Launch a Win-Back Campaign for At-Risk Customers

**Finding:** 1,537 At-Risk customers have an average spend of £726 and haven't purchased in 185 days on average. Combined with 1,986 Lost customers, the total historical revenue at stake is £2,073,816. A 20% reactivation rate at historical spend levels recovers £414,763.

**What this means:** At-Risk customers are disengaging but not gone — 185 days is long enough to be concerning but short enough that a personalised offer can still work. This window closes the longer it's left.

**Recommendation:** Run a time-sensitive win-back campaign targeting At-Risk customers first, prioritising those with spend above £500 (identified in the SQL analysis). Personalised outreach with a limited-time incentive — discount, early access, or loyalty points — within the next 30 days. For Lost customers, apply the same approach but only to those with historical spend above £1,000 where the reacquisition cost is justified.

**Target outcome:** 15–20% reactivation rate among At-Risk customers — approximately £167,000–£222,000 in recovered revenue.

## Recommendation 3 — Protect and Reward Champions

**Finding:** 1,783 Champions generate £1.9M — 37.3% of total revenue — from 22.3% of customers. They buy every 55 days on average, place 8.3 orders, and spend £1,069 each. Despite this, 47% of them hold Bronze or Silver loyalty status and are not being treated accordingly.

**What this means:** Champions are the most important customers in the business. Losing even a fraction of them to a competitor would have a disproportionate revenue impact. They are currently under-recognised by the loyalty programme, which means they have no structural reason to stay loyal beyond their own habits.

**Recommendation:** Create a dedicated Champions retention programme — exclusive early access to new products, personalised outreach, VIP event invitations, and direct acknowledgement of their value to the business. Upgrade all Champions currently sitting in Bronze or Silver to at least Gold status immediately. Monitor Champion churn rate as a primary KPI going forward.

**Target outcome:** Reduce Champion churn rate and maintain revenue concentration above 35% from this segment.

## Recommendation 4 — Invest in Beauty & Health

**Finding:** Beauty & Health has the highest margin of any category at 51.4%, but it is significantly underrepresented in revenue relative to its profitability. Electronics drives 54% of revenue at only 27.7% margin — more than double the revenue contribution of Beauty & Health at less than half the margin.

**What this means:** The business is heavily dependent on a low-margin category. Beauty & Health generates far more profit per pound of revenue than Electronics, but is not being marketed or promoted at the same level. Every pound shifted from Electronics to Beauty & Health in the revenue mix improves overall profitability.

**Recommendation:** Increase promotional activity, shelf/digital space, and cross-sell placement for Beauty & Health — particularly targeting Champions and Loyal customers who are already high-frequency buyers. Bundle Beauty & Health products with Electronics purchases through the Mobile App channel, which has the highest AOV (£111.85) and is most likely to drive basket size increases.

**Target outcome:** Increase Beauty & Health revenue share from its current position by 3–5 percentage points within 12 months, improving overall blended margin.

## Recommendation 5 — Prioritise the Mobile App Channel

**Finding:** The Mobile App has the highest average order value of any channel at £111.85, ahead of both In-Store and Online. It is the channel where customers spend the most per transaction.

**What this means:** Mobile App users are the most valuable channel customers by transaction size. Any investment in the app — features, personalisation, offers — has the highest revenue-per-user ceiling of the three channels.

**Recommendation:** Prioritise Mobile App development and promotional investment. Push Champions and Loyal customers toward the app through exclusive app-only offers and early access deals. Use the app as the primary channel for win-back campaigns — push notifications are more immediate and personal than email for re-engagement.

**Target outcome:** Increase Mobile App's share of total transactions while maintaining or growing its AOV advantage over other channels.

## Recommendation 6 — Convert New Customers Before They Go Quiet

**Finding:** 652 New Customers (8.2% of the base) have an average recency of 61 days and an average spend of £358. They are recent buyers but low frequency — 3.2 orders on average. Without intervention, a significant portion of these customers will slide into Needs Attention or Lost.

**What this means:** The first repeat purchase is the critical conversion moment. Customers who buy twice are significantly more likely to buy a third time. New Customers who don't receive a prompt within 30–60 days of their first purchase have a much higher probability of lapsing.

**Recommendation:** Implement an automated second-purchase incentive triggered 21–30 days after a first purchase — discount on a complementary category, loyalty points bonus, or free shipping threshold. Prioritise customers whose first purchase was in a high-margin category like Beauty & Health, where the second purchase economics are strongest.

**Target outcome:** Increase the New Customer → Loyal conversion rate, reducing the volume of customers who lapse before a second purchase.
